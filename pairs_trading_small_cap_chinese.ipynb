```python
import pairs_trading_strategy_lib as pts
tickers= ["JFIN", "YMM", "BZ", "ZH", "DDL", "WKEY", "WAFU", "MOMO", "WB", "FENG"]
data = pts.get_historical_data(tickers, "6mo")
pts.create_correlation_heatmap(data)
```

```python
pairs = [("ZH", "WB"), ("DDL", "WB"), ("JFIN", "ZH"), ("JFIN", "WB"), ("ZH", "WAFU")]
for pair in pairs:
    pts.plot_ratio(data, pair)
```

```python
zh_wb = ("ZH", "WB")
pairs = [("ZH", "WB"), ("JFIN", "WB"), ("JFIN", "ZH")]
for pair in pairs:
    pts.cointegration_test(data, pair)
    ratio = pts.plot_ratio(data, pair)
    z_score = pts.plot_zscore(ratio, [-0.75, -1, 0.75, 1], pair)
    pts.develop_strategy(ratio, "Ratio", z_score, -1, 0.75, pair)
```
